# TITANIC SURVIVAL PREDICTION

## IMPORT LIBRARIES

```python
import pandas as pd 
import matplotlib.pyplot as plt 
import numpy as np
```

## importing the dataset

```python
train_set = pd.read_csv("train.csv")
train_set.head()
```

```python
train_set = train_set.drop(["PassengerId","Name","Ticket","Cabin"],axis=1)
```

```python
train_set
```

```python
train_set.info()
```

```python
x = train_set.iloc[:,1:8].values
y = train_set.iloc[:,0].values
```

```python
print(x)
```

```python
print(y)
```

```python
from sklearn.impute import SimpleImputer
impute = SimpleImputer(missing_values=np.nan,strategy='mean')
x[:,2:3]= impute.fit_transform(x[:,2:3])
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test,y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=41)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
ct = ColumnTransformer(transformers=[('encoder',OneHotEncoder(handle_unknown='ignore'),[1,6])],remainder='passthrough')
x_train =np.array(ct.fit_transform(x_train))
x_test =np.array(ct.transform(x_test))
```

```python
print(x_train)
```

## selecting the model

```python
from sklearn.tree import DecisionTreeClassifier
model1 =DecisionTreeClassifier(criterion='entropy',random_state=0)
model1.fit(x_train,y_train)
```

```python
from sklearn.ensemble import RandomForestClassifier
model2 = RandomForestClassifier(n_estimators=50,criterion='entropy',random_state=0)
model2.fit(x_train,y_train)
```

## prediciting the model

```python
y_pred_tree = model1.predict(x_test)
y_pred_forest = model2.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
print("tree accoracy : ",accuracy_score(y_test,y_pred_tree))

print("forest accoracy : ",accuracy_score(y_test,y_pred_forest))
```
