## dropping the cabin column

Reason : cabin column contain more the 70% null value 

```python
train_set = train_set.drop(columns=['Cabin'])
test_set  = test_set.drop(columns=['Cabin'])
```

```python
'Cabin' in train_set.columns
```

```python
train_set['Age']=pd.to_numeric(train_set['Age'],errors='coerce')
test_set['Age']=pd.to_numeric(train_set['Age'],errors='coerce')
```

```python
age_median = train_set['Age'].median()

train_set['Age'] = train_set['Age'].fillna(age_median)
test_set['Age']  = test_set['Age'].fillna(age_median)
```

```python
type(train_set['Age'].isna().sum())
train_set.dtypes
```

```python
train_set['Age'].apply(type).value_counts()
```

using categorian feature on EMBARKED as it contain categorial data 

```python
embarked_mode = train_set['Embarked'].mode()[0]
train_set['Embarked']=train_set['Embarked'].fillna(embarked_mode)
test_set['Embarked']=test_set['Embarked'].fillna(embarked_mode)
```

```python
train_set['Embarked'].isna().sum()
```

## Encoding the categorial data

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore')
```

```python
cat_cols = ['Sex', 'Embarked']
X_train_cat = ohe.fit_transform(train_set[cat_cols])
X_test_cat = ohe.transform(test_set[cat_cols])
ohe_cols = ohe.get_feature_names_out(cat_cols)
```

```python
X_train_cat = pd.DataFrame(X_train_cat, columns=ohe_cols, index=train_set.index)
X_test_cat  = pd.DataFrame(X_test_cat,  columns=ohe_cols, index=test_set.index)
```
