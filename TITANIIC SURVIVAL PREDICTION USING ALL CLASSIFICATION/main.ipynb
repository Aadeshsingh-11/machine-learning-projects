## dropping the cabin column

Reason : cabin column contain more the 70% null value 

```python
train_set = train_set.drop(columns =['Cabin'])
test_set = test_set.drop(columns=['Cabin'])
```

```python
train_set['Age']=pd.to_numeric(train_set['Age'],errors='coerce')
test_set['Age']=pd.to_numeric(train_set['Age'],errors='coerce')
```

```python
age_median = train_set['Age'].median()

train_set['Age'] = train_set['Age'].fillna(age_median)
test_set['Age']  = test_set['Age'].fillna(age_median)
```

```python
embarked_mode = train_set['Embarked'].mode()[0]
train_set['Embarked']=train_set['Embarked'].fillna(embarked_mode)
```
