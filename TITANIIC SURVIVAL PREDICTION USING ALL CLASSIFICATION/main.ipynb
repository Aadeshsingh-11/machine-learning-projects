# TITANIC SURVIVAL PREDICTION

## IMPORT LIBRARIES

```python
import pandas as pd 
import matplotlib.pyplot as plt 
import numpy as np
```

## importing the dataset

```python
train_set = pd.read_csv("train.csv")
train_set.head()
```

```python
train_set = train_set.drop(["PassengerId","Name","Ticket","Cabin"],axis=1)
```

```python
train_set
```

```python
train_set.info()
```

```python
x = train_set.iloc[:,1:8].values
y = train_set.iloc[:,0].values
```

```python
print(x)
```

```python
print(y)
```

```python
from sklearn.impute import SimpleImputer
impute = SimpleImputer(missing_values=np.nan,strategy='mean')
x[:,2:3]= impute.fit_transform(x[:,2:3])
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test,y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=41)
```

## applying feature scalling

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train[:,[2,5]] = sc.fit_transform(x_train[:,[2,5]])
x_test[:,[2,5]] = sc.transform(x_test[:,[2,5]])
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
ct = ColumnTransformer(transformers=[('encoder',OneHotEncoder(handle_unknown='ignore'),[1,6])],remainder='passthrough')
x_train =np.array(ct.fit_transform(x_train))
x_test =np.array(ct.transform(x_test))
```

```python
print(x_train)
```

## selecting the model

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(random_state = 0)
model.fit(x_train,y_train)
```

```python
from sklearn.svm import SVC
model1 = SVC(kernel='rbf',random_state=0)
model1.fit(x_train,y_train)
```

```python
from sklearn.neighbors import KNeighborsClassifier
model2=KNeighborsClassifier(n_neighbors=5,weights='uniform',metric='minkowski',p=2)
model2.fit(x_train,y_train)
```

```python
from sklearn.naive_bayes import GaussianNB
model3 = GaussianNB()
model3.fit(x_train,y_train)
```

## prediciting the model

```python
y_pred_logistic = model.predict(x_test)
y_pred_svm = model1.predict(x_test)
y_pred_neighbour =model2.predict(x_test)
y_pred_naive = model3.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
print("svm accoracy : ",accuracy_score(y_test,y_pred_svm))
print("neighbour accoracy : ",accuracy_score(y_test,y_pred_neighbour))
print("logistic accoracy : ",accuracy_score(y_test,y_pred_logistic))
print("naive bayes accuracy : ", accuracy_score(y_test, y_pred_naive))
```
