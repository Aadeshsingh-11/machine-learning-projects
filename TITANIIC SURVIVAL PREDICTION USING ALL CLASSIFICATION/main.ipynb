## importing the dataset

```python
train_set = pd.read_csv("train.csv")
train_set.head()
```

```python
train_set = train_set.drop(["PassengerId","Name","Ticket","Cabin"],axis=1)
```

```python
x = train_set.iloc[:,1:8].values
y = train_set.iloc[:,0].values
```

```python
from sklearn.impute import SimpleImputer
impute = SimpleImputer(missing_values=np.nan,strategy='mean')
impute.fit_transform(x[:,2:3])
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
ct = ColumnTransformer(transformers=[('encoder',OneHotEncoder(),[1,6])],remainder='passthrough')
x =ct.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test,y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=41)
```
